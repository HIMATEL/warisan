c. Ubah frekuensi sampling menjadi 8 Hz, apa yang terjadi?

```python
f=4                                                        # frekuensi
dur = 1                                                    # panjang/durasi dari sinyal (sepanjang sumbu x)
fs=8                                                     # frekuensi sampling
fasa_awal=0                                                # fasa awal, jika ditulis sebagai derajat maka perlu dikonversikan ke radian
A=1                                                        # amplitudo
B=0                                                        # offset
t = np.linspace(0,dur,fs*dur)                  # x
s = B+A*np.sin(2*np.pi*f*t+np.deg2rad(fasa_awal))  # y

plt.plot(t,s)
# plt.xlabel('Time (s)')
# plt.ylabel('Voltage (mV)')
# plt.title('Sine Plot')
# plt.xlim(0,0.25)
# plt.ylim(-1.1,1.1)
# plt.yticks(set_ticks_array(-1,1,0.5))
# plt.xticks(set_ticks_array(0,0.25,0.05))
plt.grid()

plt.show()
```

Ini berbeda dari sebelumnya, di mana dengan frekuensi sampling 100 Hz, sinyal terlihat halus dan menyerupai gelombang sinus yang asli.


d. Berapa frekuensi sampling minimal untuk mendapatkan jumlah gelombang yang sama dengan frekuensi-nya. Dan mengapa?


```python
f=4                                                        # frekuensi
dur = 1                                                    # panjang/durasi dari sinyal (sepanjang sumbu x)
fs=f*10                                                  # frekuensi sampling
fasa_awal=0                                                # fasa awal, jika ditulis sebagai derajat maka perlu dikonversikan ke radian
A=1                                                        # amplitudo
B=0                                                        # offset
t = np.linspace(0,dur,fs*dur)                  # x
s = B+A*np.sin(2*np.pi*f*t+np.deg2rad(fasa_awal))  # y

plt.plot(t,s)
# plt.xlabel('Time (s)')
# plt.ylabel('Voltage (mV)')
# plt.title('Sine Plot')
# plt.xlim(0,0.25)
# plt.ylim(-1.1,1.1)
# plt.yticks(set_ticks_array(-1,1,0.5))
# plt.xticks(set_ticks_array(0,0.25,0.05))
plt.grid()

plt.show()
```

Dengan menganalisis soal sebelumnya, dapat disimpulkan bahwa semakin besar nilai frekuensi sampling, maka sinyal yang tampil akan memiliki sinyal sinus yang halus dan menyerupai hasilnya.

Pada soal ini pertama dicoba untuk frekuensi sampling dua kali lipat lebih besar dari nilai frekuensi, maka hasilnya sinyal tidak dapat ditampilkan dengan baik, sinyalnya patah-patah dan jumlah gelombang yang tampil tidak sama dengan besar frekuensinya.

Kemudian, dicoba untuk frekuensi sampling sepuluh kali lipat dari nilai frekuensi.Sinyal yang dihasilkan memang belum halus dan masih terdapat patah-patah, namun jumlah gelombang yang dihasilkan sudah sama seperti besar frekuensi yag ditentukan. Sepuluh kali lipat dari frekuensi adalah nilai minimal untuk mendapatkan jumlah gelombang yang sama dengan nilai frekuensi, namun masih menghasilkan bentuk sinyal yang tidak halus seperti bentu sinus dan sinyal aslinya.

f. Dari program di atas, lakukan modifikasi agar menghasilkan Audio dimana frekuensinya berubah seiring durasi (bisa frekuensinya terus naik, atau terus turun, atau bersifat fluktuatif) dengan frekuensi yang digunakan diantara 200 Hz s/d 4000 Hz.

```python
dur = 5
fs = 8000
fasa_awal = 0
A = 1
B = 0
t = np.linspace(0, dur, int(fs * dur))

f_t = 200 + (4000 - 200) * (t / dur)

s = B + A * np.sin(2 * np.pi * np.cumsum(f_t) / fs + np.deg2rad(fasa_awal))
s = s / np.max(np.abs(s))

Audio(s, rate=fs)
```
