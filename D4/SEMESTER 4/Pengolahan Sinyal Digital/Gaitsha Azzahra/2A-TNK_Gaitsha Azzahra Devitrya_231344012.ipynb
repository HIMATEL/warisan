```python
import warnings
warnings.filterwarnings("ignore")
import numpy as np
from scipy import signal as sgn
import matplotlib
import matplotlib.pyplot as plt
def set_ticks_array(minVal,maxVal,incVal):
    ticks_array=np.linspace(minVal,maxVal, int((maxVal-minVal)/incVal+1),endpoint=True)
    return ticks_array
frekuensi=8
frekuensi2=4
frekuensi_sampling=100
fasa_awal=0
A=1
B=0
t = np.linspace(0,0.5,frekuensi_sampling)
s = B+A*np.sin(2*np.pi*frekuensi*t+np.deg2rad(fasa_awal))
s2= B+A*np.sin(2*np.pi*frekuensi2*t+np.deg2rad(fasa_awal))

plt.plot(t,s+s2)
plt.xlabel('Time (s)')
plt.ylabel('Voltage (mV)')
plt.title('Sine Plot')
plt.xlim(-0.02,0.52)
plt.ylim(-2,2)
plt.yticks(set_ticks_array(-1.5,1.5,0.5))
plt.xticks(set_ticks_array(0,0.5,0.1))
plt.grid()

plt.show()
```

> *Jelaskan Program Anda Disini*


Percobaan soal ke-1, program ini membuat dan menampilkan grafik hasil penjumlahan dua sinyal sinusoidal dengan frekuensi 8 Hz dan 4 Hz menggunakan Numpy. Di deklarasikan terlebih dahulu parameter seperti amplitudo, fase awal, dan frekuensi sampling, program membuat vektor waktu t dan menghitung dua sinyal sinusoidal s dan s2. Kedua sinyal dijumlahkan dan diplot dengan label sumbu, batas skala, serta garis bantu grid. Fungsi set_ticks_array digunakan untuk mengatur tanda skala sumbu x dan y, memastikan tampilan grafik lebih terstruktur. Akhirnya, grafik ditampilkan menggunakan plt.show().

Pada soal ke-2, program ini membuat grafik sinyal gigi gergaji menggunakan NumPy, SciPy, dan Matplotlib. Pertama, program mengatur parameter seperti frekuensi, frekuensi sampling, fasa awal, amplitudo (A), dan offset (B). Variabel t berisi nilai waktu, sedangkan sinyal s dihitung menggunakan fungsi sawtooth dari SciPy.

Grafik ditampilkan dengan stem plot, di mana sumbu x mewakili waktu dan sumbu y menunjukkan tegangan. Rentang sumbu diatur dengan plt.xlim() dan plt.ylim(), serta tick disesuaikan menggunakan fungsi set_ticks_array(). Grid juga ditambahkan untuk memperjelas tampilan.

```python
frekuensi=1/20
frekuensi_sampling= 50
fasa_awal=0                                            # fasa awal, jika ditulis sebagai derajat maka perlu dikonversikan ke radian
A=2                                                     # amplitudo
B=0
width=0.5                                                   # offset
t = np.linspace(1,50,frekuensi_sampling)                 # x
s = B+A*sgn.sawtooth(2*np.pi*frekuensi*t+np.deg2rad(fasa_awal), width)  # y


plt.stem(t, s)
plt.xlabel('Time (s)')
plt.ylabel('Voltage (mV)')
plt.title('Sine Plot')
plt.xlim(0,0.5)
plt.ylim(-2.1,2.1)
plt.yticks(set_ticks_array(-2,2,1))
plt.xticks(set_ticks_array(0,50,10))
plt.grid()

plt.show()
```

> *Jelaskan Program Anda Disini*

Program ini membuat grafik gelombang gigi gergaji menggunakan NumPy dan Matplotlib. Frekuensi sinyal ditetapkan 1/20 Hz, dengan frekuensi sampling 50 Hz, amplitudo 2, dan fase awal 0 derajat. Array waktu t dibuat dengan np.linspace(), lalu sinyal s dihitung menggunakan sgn.sawtooth(), yang menghasilkan gelombang periodik dengan lebar naik 0.5 periode.

Grafik ditampilkan menggunakan plt.stem(t, s), dengan label sumbu, judul, serta batas sumbu x (0–0.5 s) dan y (-2.1–2.1). Tick sumbu x dan y diatur agar mudah dibaca, dan grid ditambahkan untuk memperjelas tampilan. Akhirnya, plt.show() digunakan untuk menampilkan plot.

```python
frekuensi=1/24
frekuensi_sampling= 22
fasa_awal=8
A=14.5
B=0
t = np.linspace(0,23,frekuensi_sampling)
s = B+A * np.sin (2*np.pi*frekuensi*t+np.deg2rad(fasa_awal))
s2= B+A * sgn.square(2*np.pi*frekuensi*t+np.deg2rad(fasa_awal))

plt.figure(figsize=(10,7))
plt.stem(t,s2-s)
plt.xlabel('Time (s)')
plt.ylabel('Voltage (mV)')
plt.title('Sine Plot')
#plt.xlim(0,25)
#plt.ylim(-15,15)
plt.yticks(set_ticks_array(-15,15,5))
plt.xticks(set_ticks_array(0,25,5))
plt.grid()

plt.show()
```

> *Jelaskan Program Anda Disini*

Program ini membuat plot perbedaan antara sinyal sinusoidal dan sinyal kotak menggunakan Python dan Matplotlib. Pertama, parameter seperti frekuensi (1/24), frekuensi sampling (22), amplitudo (14.5), dan fasa awal (0) ditentukan. Kemudian, sinyal sinusoidal (s) dan sinyal kotak (s2) dihitung menggunakan fungsi sinus dan square wave.

Plot batang dibuat menggunakan plt.stem(t, s2 - s), dengan label sumbu X (Time) dan Y (Voltage) serta judul Sine Plot. Batas sumbu, skala tick, dan grid ditambahkan untuk memperjelas visualisasi. Akhirnya, plt.show() digunakan untuk menampilkan grafik.
