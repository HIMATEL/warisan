### Pendahuluan

**Elias Gamma Coding**

Metode ini dikembangkan oleh Peter Elias. Dengan teknik pengkodean:
1. Tulis dalam biner
2. Jumlah bit pada proses no.1 dikurang satu untuk menentukan jumlah nol (zeros) yang
harus ditambahkan di muka angka tersebut.
Contoh ada sebuat bilangan 11:
1. Langkah pertama adalah konversi 11 menjadi biner, yakni 1011.
2. Cari jumlah bit/pangkat tertinggi pada proses pertama, lalu kurangi 1.

Gunakan jumlah tersebut untuk menambahkan angka 0 didepan biner yang terdapat pada proses pertama.
Hasilnya adalah 0001011.

Sehingga jika didapatkan deretan angka 11, 15, dan 27 maka dengan elias gamma code deretan angka tersebut dapat diubah menjadi 00010110001111000011011.

Fungsi dari elias gamma code ini adalah pengguna dapat mengirim banyak angka berjajar sekaligus, sebab jika kita hanya mengubah bilangan tersebut menjadi biner maka akan sulit melakukan decode.

Sebagai contoh tiga bilangan yang terdapat pada contoh di atas. Jika data
tersebut dikirim 1011111111011 saja, maka akan ada banyak variasi kombinasi biner untuk mendapatkan nilai desimalnya.

**Elias Delta Coding**

Seperti Elias Gamma, kode ini ditemukan oleh Peter Elias. Kode ini menerapkan juga metode
pada gamma coding, terutama di bagian kepala. Tekniknya adalah sebagai berikut:
1. Cari pangkat tertinggi binernya, misalnya desimal 11 jika dibinerkan menjadi 1011 dimana
pangkat tertingginya adalah 3. Jadi N’=3.
2. Gunakan Gamma Coding untuk mengkodekan bilangan N dimana N=N’+1. Jadi untuk
kasus desimal 11 maka kita harus membuat Gamma Coding dari 4 yaitu 00100.
3. Tambahkan sisa N’ binary pada hasil no.2. Jadi diperoleh jawaban 00100011.
Jika diberikan deretan angka 35, 101, dan 112, maka akan didapatkan output dengan metode Elias Delta Code sebagai berikut:

00110000110011110010100111110000

2. Gunakan data yang sama pada nomor satu untuk encoder dengan metode Elias Delta Code.

**Instruksi:**





```python
#tuliskan fungsi program pada tiap barisnya

def elliasGamma(input): #untuk apa perintah def? perintah def adalah kata kunci yang digunakan untuk mendefinisikan sebuah fungsi. Ini mendefinisikan sebuah fungsi bernama elliasGamma dengan satu parameter input.
 gammaResult = "" #apa fungsi variabel gammaResult? variabel gammaResult ini adalah deklarasi variabel gammaResult, yang awalnya diinisialisasi sebagai string kosong.
 biner = format(input,'b') #untuk apa perintah format? Perintah ini mengonversi bilangan input ke bentuk biner dan menyimpannya dalam variabel biner. format(input, 'b') digunakan untuk mengonversi input ke bilangan biner.

 for n in range(len(biner)-1): #apa tujuan dari for ini? tujuannya adalah loop for yang digunakan untuk menghasilkan sejumlah '0' sesuai dengan panjang biner input dikurangi satu.
  gammaResult = '0' + gammaResult

 gammaResult = gammaResult + biner # Ini menggabungkan hasil dari loop sebelumnya dengan nilai biner asli dan menyimpannya kembali dalam variabel gammaResult.

 return gammaResult #untuk apa perintah return? Ini adalah pernyataan return yang mengembalikan nilai gammaResult sebagai output dari fungsi elliasGamma.

#----------------------------------------------

def elliasDelta(input): #untuk apa perintah def? ini adalah definisi fungsi yang mendefinisikan fungsi elliasDelta dengan satu parameter input.
 biner = format(input,'b') #untuk apa perintah format? ini mengonversi bilangan input ke bentuk biner dan menyimpannya dalam variabel biner.
 n = len(biner) #apa fungsi variabel n ? Ini menghitung panjang biner yang dihasilkan dan menyimpannya dalam variabel n.

 gamma = elliasGamma(n) #apa fungsi variabel gamma ? Ini memanggil fungsi elliasGamma untuk mengkodekan panjang biner n dan menyimpan hasilnya dalam variabel gamma.
 deltaResult = gamma + biner[1:] #apa fungsi variabel deltaResult? untuk apa biner[1:] ? Ini menggabungkan hasil dari Elias Gamma (variabel gamma) dengan biner yang dimulai dari indeks kedua (indeks 1) hingga akhir (indeks terakhir). Ini menghasilkan hasil Elias Delta dan menyimpannya dalam variabel deltaResult.

 return deltaResult #untuk apa perintah return? Ini adalah pernyataan return yang mengembalikan nilai deltaResult sebagai output dari fungsi elliasDelta.

#----------------------------------------------

data = [23, 35, 36, 44, 45] #untuk apa variabel data? Variabel data adalah sebuah daftar (list) yang berisi sejumlah angka. Ini digunakan sebagai input untuk kedua fungsi elliasGamma dan elliasDelta.
out_gamma = ''
out_delta = ''

for i in data:
 out_gamma = out_gamma + ' ' + elliasGamma(i) #untuk apa variabel out_gamma? Ini adalah loop for yang digunakan untuk mengiterasi melalui setiap elemen dalam daftar data, mengkodekannya menggunakan fungsi elliasGamma dan menyimpan hasilnya dalam variabel out_gamma
 out_delta = out_delta + ' ' + elliasDelta(i) #untuk apa variabel out_delta? Ini adalah loop for yang digunakan untuk mengiterasi melalui setiap elemen dalam daftar data, mengkodekannya menggunakan fungsi elliasDelta dan menyimpan hasilnya dalam variabel out_delta

print('input gamma: \n\t',data)
print('output gamma: \n\t',out_gamma)
print('input delta: \n\t',data)
print('output delta: \n\t',out_delta)
```

Variabel data berisi kumpulan angka yang akan diolah oleh kedua fungsi di atas. Variabel out_gamma dan out_delta digunakan untuk menyimpan hasil dari fungsi elliasGamma dan elliasDelta untuk setiap angka dalam data.

Terakhir, program mencetak hasil keluaran dari kedua fungsi untuk data dalam format yang sesuai dengan judulnya. Jadi, program ini adalah implementasi dari metode kompresi data Elias Gamma dan Elias Delta yang digunakan untuk mengkodekan angka menjadi urutan bit yang lebih efisien.
