### Pendahuluan

**Elias Gamma Coding**

Metode ini dikembangkan oleh Peter Elias. Dengan teknik pengkodean:
1. Tulis dalam biner
2. Jumlah bit pada proses no.1 dikurang satu untuk menentukan jumlah nol (zeros) yang
harus ditambahkan di muka angka tersebut.
Contoh ada sebuat bilangan 11:
1. Langkah pertama adalah konversi 11 menjadi biner, yakni 1011.
2. Cari jumlah bit/pangkat tertinggi pada proses pertama, lalu kurangi 1.

Gunakan jumlah tersebut untuk menambahkan angka 0 didepan biner yang terdapat pada proses pertama.
Hasilnya adalah 0001011.

Sehingga jika didapatkan deretan angka 11, 15, dan 27 maka dengan elias gamma code deretan angka tersebut dapat diubah menjadi 00010110001111000011011.

Fungsi dari elias gamma code ini adalah pengguna dapat mengirim banyak angka berjajar sekaligus, sebab jika kita hanya mengubah bilangan tersebut menjadi biner maka akan sulit melakukan decode.

Sebagai contoh tiga bilangan yang terdapat pada contoh di atas. Jika data
tersebut dikirim 1011111111011 saja, maka akan ada banyak variasi kombinasi biner untuk mendapatkan nilai desimalnya.

**Elias Delta Coding**

Seperti Elias Gamma, kode ini ditemukan oleh Peter Elias. Kode ini menerapkan juga metode
pada gamma coding, terutama di bagian kepala. Tekniknya adalah sebagai berikut:
1. Cari pangkat tertinggi binernya, misalnya desimal 11 jika dibinerkan menjadi 1011 dimana
pangkat tertingginya adalah 3. Jadi N’=3.
2. Gunakan Gamma Coding untuk mengkodekan bilangan N dimana N=N’+1. Jadi untuk
kasus desimal 11 maka kita harus membuat Gamma Coding dari 4 yaitu 00100.
3. Tambahkan sisa N’ binary pada hasil no.2. Jadi diperoleh jawaban 00100011.
Jika diberikan deretan angka 35, 101, dan 112, maka akan didapatkan output dengan metode Elias Delta Code sebagai berikut:

00110000110011110010100111110000

### Langkah-langkah Percobaan

1. Analisis dan pelajari program encoder dengan metode Elias Gamma Code untuk data 23, 35, 36, 44, 45.

2. Gunakan data yang sama pada nomor satu untuk encoder dengan metode Elias Delta Code.

**Instruksi:**





```python
#tuliskan fungsi program pada tiap barisnya

def elliasGamma(input): #untuk apa perintah def? Def digunakan untuk mendefinisikan fungsi bernama elliasGamma. Fungsi ini untuk mengubah angka menjadi kode elias gamma
 gammaResult = "" #apa fungsi variabel gammaResult? Variabel untuk menyimpan hasil sementara dari proses encoding elias gamma. Pada awal variabel ini diisi dengan string kosong
 biner = format(input,'b') #untuk apa perintah format? Mengubah angka input menjadi bentuk biner,sistem bilangan dengan angka 0 dan 1. Jika input adalah 5 maka hasilnya 101

 for n in range(len(biner)-1): #apa tujuan dari for ini? Perulangan ini berjalan sebanyak (len(biner) - 1) kali (yaitu panjang string biner dikurangi 1). Tujuannya adalah menambahkan angka nol (0) ke awal gammaResult. Jika biner 101 jumlah bitnya adalah 3 maka berulangan ini berjalan 2kali (3-1) maka hasil akhirnya menjadi 00
  gammaResult = '0' + gammaResult

 gammaResult = gammaResult + biner #setelah menambahkan nol, biner asli digabungkan di belakang hasil sebelum proses ini contohnya 00+101 menjadi 00101

 return gammaResult #untuk apa perintah return?Mengirim hasil encoding gammaResult ke bagian lain program yang memanggil fungsi ini.

#----------------------------------------------

def elliasDelta(input): #untuk apa perintah def?Mendefinisikan fungsi untuk encoding dengan metode Elias Delta.
 biner = format(input,'b') #untuk apa perintah format?Untuk Mengubah angka input menjadi string biner. Contohnya jika input adalah 9 maka hasilnya 1001
 n = len(biner) #apa fungsi variabel n, untuk menyimpan panjang dari string biner. Contohnya 1001 memiliki panjang 4, sehingga n = 4

 gamma = elliasGamma(n) #apa fungsi variabel gamma? Untuk Memanggil fungsi Elias Gamma untuk menghasilkan kode dari panjang biner (n). Contohnya jika n=4, hasil elliasGamma(4)adalah 00100
 deltaResult = gamma + biner[1:] #apa fungsi variabel deltaResult? untuk apa biner[1:] = deltaResult: Menyimpan hasil encoding Elias Delta.
 #Kode Gamma dari panjang biner (gamma) digabungkan dengan biner tanpa digit pertama (biner[1:]).
 #Contoh: '1001' → tanpa digit pertama → '001'. Jadi hasil akhirnya adalah '00100' + '001' = '00100001'.

 return deltaResult #untuk apa perintah return? Untuk Mengirim hasil encoding Elias Delta (deltaResult) ke bagian lain program.

#----------------------------------------------

data = [23, 35, 36, 44, 45] #untuk apa variabel data? Data ini adalah list berisi angka-angka yang akan di-encode menggunakan metode Elias Gamma dan Elias Delta.. Contohnya 23,35,36,44,45
out_gamma = '' #digunakan untuk menyimpan hasil encoding Gamma dan Delta dari semua angka dalam list data.
out_delta = '' #digunakan untuk menyimpan hasil encoding Gamma dan Delta dari semua angka dalam list data.

for i in data:
 out_gamma = out_gamma + ' ' + elliasGamma(i) #untuk apa variabel out_gamma? Menyimpan semua hasil encoding Elias Gamma.
 out_delta = out_delta + ' ' + elliasDelta(i) #untuk apa variabel out_delta? Menyimpan semua hasil encoding Elias Delta.
#contoh: Angka 23 → Gamma: '00010111', Delta: `'00100111

#Menampilkan hasil input dan output dari encoding Elias Gamma dan Elias Delta.
print('input gamma: \n\t',data) #Menampilkan daftar angka yang diprosesdari input gamma.
print('output gamma: \n\t',out_gamma) #Menampilkan hasil akhir encoding gamma.
print('input delta: \n\t',data)#Menampilkan daftar angka yang diproses dari input delta.
print('output delta: \n\t',out_delta) #Menampilkan hasil akhir encoding delta.
```



> Analisa dari program di atas:

Program ini mengimplementasikan dua metode encoding yang efisien, yaitu Elias Gamma dan Elias Delta, yang digunakan untuk mengonversi angka desimal menjadi representasi biner yang lebih ringkas. Fungsi `elliasGamma` memanfaatkan format biner untuk menentukan panjang dari angka dan menambahkan nol di depan string hasil encoding untuk menghasilkan format yang sesuai. Selanjutnya, fungsi `elliasDelta` memanggil fungsi gamma untuk mendapatkan kode dari panjang biner dan menggabungkannya dengan biner input tanpa digit pertama. Pendekatan ini memastikan bahwa hasil encoding menjadi lebih kompak, sehingga mengurangi penggunaan ruang penyimpanan.

Dalam bagian utama program, daftar angka diolah melalui loop untuk menghasilkan hasil encoding Gamma dan Delta, yang kemudian disimpan dalam variabel terpisah. Setelah semua angka diproses, hasilnya ditampilkan dengan jelas, memberikan gambaran yang komprehensif tentang input dan output dari kedua metode encoding tersebut.
