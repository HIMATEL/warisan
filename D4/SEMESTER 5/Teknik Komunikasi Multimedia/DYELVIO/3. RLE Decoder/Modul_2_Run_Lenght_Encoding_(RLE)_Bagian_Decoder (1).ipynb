### Pendahuluan

Dalam usaha untuk mengirimkan data dari satu tempat ke tempat lain, seringkali terjadi ketidakseimbangan antara jumlah data yang akan dikirim dengan kapasitas jaringan yang tersedia. Akibatnya, kualitas layanan dalam sistem pengiriman data menjadi kurang memadai. Salah satu cara untuk mengatasi situasi ini adalah dengan melakukan kompresi data sebelum pengiriman, sehingga beban pada jaringan dapat berkurang. Teknik kompresi data dapat diterapkan jika terdapat pola berulang dalam data, di mana pola-pola ini dapat diwakili dengan singkat pada awalnya dan kemunculan selanjutnya hanya merujuk pada pola awal tersebut. Terdapat berbagai teknik kompresi data yang sudah dikembangkan, dan pada kesempatan ini akan diterapkan metode kompresi data yang dikenal sebagai Run Length Encoding (RLE). RLE merupakan teknik kompresi data yang mengompresikan pola yang muncul secara berurutan. Contoh penggunaan RLE adalah dengan mengganti setiap rangkaian angka 0 dengan 2 karakter, yang menunjukkan bahwa kompresi sedang dilakukan, diikuti dengan jumlah angka 0 dalam rangkaian tersebut. Semakin banyak angka 0 yang berurutan, maka ukuran file hasil kompresi akan semakin efisien.

Berdasarkan pada program RLE Encoder sebelumnya, algoritma program RLE Decoder seharusnya tidak jauh berbeda dari program sebelumnya. Di program Encoder mahasiswa sudah dapat membuat program untuk mencari suatu simbol yang akan diencode (atau dalam kasus sebelumnya mencari angka 0), kemudian saat simbol tersebut akan dicek ke simbol berikutnya apakah masih berulang atau tidak. Jika ditemukan simbol berulang maka simbol tersebut akan dihitung, sebaliknya jika tidak ada perulangan maka program hanya mencetak ulang simbol dari data stream.

Pada bagian decoder ini, mahasiswa diminta untuk bisa membuat program yang dapat mencari simbol 0. Kemudian setelah simbol 0 ditemukan, program harus bisa melihat angka didepan simbol 0 tersebut. Angka tersebut merupakan jumlah perulangan dari angka 0 yang akan didecode. Contoh jika didapatkan;

**datastream** = 1 2 3 0 5 8 9

maka setelah ditemukan angka 0, output harus bisa mencetak ulang angka 0 sebanyak angka didepannya yaitu 5. Sehingga data output yang dihasilkan menjadi;

**output** = 1 2 3 0 0 0 0 0 8 9

Untuk bisa mendapatkan ouput tersebut bisa didapat dengan sedikit melakukan modifikasi terhadap program pada praktikum sebelumnya.

1. Buatlah program untuk melakukan decoding RLE dengan mencari angka 0 lalu mengulang akan 0 tersebut sebanyak jumlah angka yang ada setelahnya.



```python
#tuliskan program yang telah dimodifikasi di sini
#buktikan dengan datastream = 4 5 1 3 2 5 3 4 5 0 7 7 6 2 3 1 4 5
#tuliskan fungsi program pada tiap barisnya

c = 0 #untuk inisialisasi nilai c dengan 0 yang nantinya akan menjadi penunjuk alamat array yang akan digunakan
a = 0 #untuk inisialisai a dengan nilai 0 yang nantinya akan menjadi wadah untuk nilai konversi string datastream menjadi int
datastream = '451325345077623145' #untuk inisialisasi nilai datastream dengan '451325345077623145'

while c < len(datastream): #untuk melakukan kondisi pengulangan yang akan terjadi apabila c masih dibawah banyaknya data pada datstream
  if datastream[c]== '0': #untuk melakukan penyeleksian yang akan dieksekusi apabila datastream[c] sama dengan 0 maka akan masuk ke sub program dalam if
    a = int(datastream[c+1]) #untuk memasukan nilai konversi string ke int pada variabel datastream[c+1] ke dalam variabel a
    i = 0 #untuk inisialisasi nilai i dengan 0 yang nantinya akan menjadi sebuah up counter
    while i < a: #untuk melalukan kondisi pengulangan yang akan terus terjadi jika i lebih kecil dari a
      print('0',end="") #untuk mencetak nilai 0 tanpa membuat baris baru
      i = i+1 #untuk melakukan proses up counter dengan proses increment sebanyak satu
    c = c+2 # proses penambahan nilai dengan c awal dengan 2(diambil nilai 2 karena untuk melompati data yang merujuk untuk jumlah banyaknya proses pengulangan)
  else: #proses kondisi jika tidak memenuhi datastream[c] sama dengan 0 maka akan menjalankan program yang berada di else
    print(datastream[c],end="") #untuk mencetak nilai dari datastream[c] tanpa membuat baris baru
    c = c+1 #untuk proses penambahan nilai c awal dengan 1
```



> Analisa dari program di atas:

Pada program di bagian decoder ini terdapat sebuah pembeda dengan program di bagian encoder. Dimana di bagian decoder ini terdapat pengulangan while di di dalam pengondisian if sehingga proses pengulangan terjadi di dalam proses penyeleksian. Kemudian ada yang membedakan dengan proses penampilan nilai 0 yang disesuaikan dengan jumlahnya yaitu dimana nilai c atau alamat array ini akan ditambahkan 2, ini dikarenakan untuk melompati data yang sebenarnya hanya menunjukkan nilai banyaknya pengulangan saja, sehingga dilompat sebanyak 2 untuk masuk ke data yang benar benar harus ditampilkan atau data aslinya.

2. Buatlah program untuk melakukan decoding RLE dengan mencari angka apapun yang
mengulang. Cetak angka tersebut sebanyak perulangan yang diminta (jumlah perulangan pada program ini ditandai dengan tanda minus).



```python
#tuliskan program yang telah dimodifikasi di sini
#buktikan dengan datastream = 6 5 -2 4 5 4 -4 5 4 5 4 2 -3 0 -3 1 2 3 4 6
#tuliskan fungsi program pada tiap barisnya

c = 0 #untuk inisialisasi nilai c dengan 0 yang nantinya akan menjadi penunjuk alamat array yang akan digunakan
a = 0 #untuk inisialisai a dengan nilai 0 yang nantinya akan menjadi wadah untuk nilai konversi string datastream menjadi int
datastream = '65-2454-454542-30-312346' #untuk inisialisasi nilai datastream dengan '65-2454-454542-30-312346'

while c < len(datastream): #untuk melakukan kondisi pengulangan yang akan terjadi apabila c masih dibawah banyaknya data pada datstream
    if datastream[c] == '-': #untuk melakukan penyeleksian yang akan dieksekusi apabila datastream[c] sama dengan -(minus) maka akan masuk ke sub program dalam if
        a = int(datastream[c+1]) #untuk memasukan nilai konversi string ke int pada variabel datastream[c+1] ke dalam variabel a
        i = 1 #untuk inisialisasi nilai i dengan 0 yang nantinya akan menjadi sebuah up counter
        while i < a: #untuk melalukan kondisi pengulangan yang akan terus terjadi jika i lebih kecil dari a
            print(datastream[c-1], end="") #untuk mencetak nilai datastream[c-1] tanpa membuat baris baru
            i = i+1 #untuk melakukan proses up counter dengan proses increment sebanyak satu
        c = c+2 #proses penambahan nilai dengan c awal dengan 2(diambil nilai 2 karena untuk melompati data yang merujuk untuk jumlah banyaknya proses pengulangan)
    else: #proses kondisi jika tidak memenuhi datastream[c] sama dengan 0 maka akan menjalankan program yang berada di else
        print(datastream[c], end="") #untuk mencetak nilai dari datastream[c] tanpa membuat baris baru
        c = c+1 #untuk proses penambahan nilai c awal dengan 1
```



> Analisa dari program di atas:

Pada program ini ada yang menjadikan pembedanya yaitu dimana proses penyeleksian dalam if yang awalnya 0 pada program sebelumnya diubah menjadi (-), kemudian di saat proses pengulangan pencetakan nilai yang awalnya berisi nilai 0 diganti dengan nilai datastream[c-1] ini karena nilai untuk program ini tidak hanya untuk 0 saja sehingga diganti dengan datastream[c-1], lalu pada program ini pun jadi diketahui bahwa proses i = 1 perlu ada di dalam proses if karena agar setiap proes pengulangan untuk pencetakan nilai yang sama berulang setiap selesai maka nilai i akan direset kembali menjadi 1, dan pembedanya pun yaitu i yang diinisialisasi dengan 1 karena nilai yang awal sudah ditampilkan maka proses inisialisasi nilai i ditambah 1 agar sesuai dengan data asli dengan hasil dari decoder
