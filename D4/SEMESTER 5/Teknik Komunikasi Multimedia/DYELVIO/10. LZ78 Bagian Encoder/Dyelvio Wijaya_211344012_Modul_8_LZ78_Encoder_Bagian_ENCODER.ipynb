**Pendahuluan**   

LZ78 adalah sebuah metode kompresi data yang dikembangkan oleh Abraham Lempel dan Jacob Ziv pada tahun 1978. Nama LZ78 berasal dari inisial kedua penemu tersebut dan tahun penemuan. Metode ini merupakan salah satu dari sejumlah algoritma kompresi data yang dikenal sebagai algoritma LZ (Lempel-Ziv).

Konsep dasar dari LZ78 adalah menggantikan pola-pola data yang sering muncul dengan simbol-simbol yang lebih pendek. Algoritma ini memanfaatkan kamus atau tabel untuk menyimpan dan mengelola pola-pola yang telah muncul sebelumnya selama proses kompresi. Setiap kali sebuah pola baru ditemukan, algoritma menyimpan pola tersebut dalam kamus dan menggantinya dengan sebuah simbol yang lebih pendek.

Proses kompresi LZ78 melibatkan dua tindakan utama: pencarian kamus untuk menemukan pola yang cocok dan menambahkan pola baru ke dalam kamus. Informasi mengenai pola-pola tersebut kemudian dijadikan sebagai referensi untuk menggantikan pola-pola serupa yang muncul selama proses kompresi.

Meskipun LZ78 merupakan salah satu metode kompresi yang cukup efisien, terdapat berbagai varian dan pengembangan lebih lanjut seperti LZ77 yang juga populer dalam implementasi kompresi data.

Encode (i.e., compress) data stream BABAABRRRA menggunakan algoritma LZ78.

### Langkah-langkah Percobaan

Buatlah program untuk Teknik kompresi LZ78 Bagian Encoder dengan input teks sebagai berikut:  BABAABRRRA  Dengan kaidah LZ78 maka hasil dari encodernya adalah:  0B, 0A, 1A, 2B, 0R, 5R, 2

Tampilkan juga dictionary yang dihasilkan dari proses LZ78 tersebut.  

LZ78 colab2.png



**Instruksi:**





```python
#tuliskan fungsi program pada tiap barisnya

data = 'BABAABRRRA' #proses inisialisasi data dengan nilai 'BABAABRRRA'
kamus = [] #proses deklarasi kamus dengan tipe data array
hasilencode = [] #proses deklarasi hasilencode dengan tipe data array
count = len(data) #proses inisialisasi count dengan banyaknya jumlah data
j=0 #proses inisialisasi variabel j dengan 0

for i in range(count): #proses perulangan dengan sebanyak jumlah data
  i-j #proses pengurangan nilai i dengan j
  #print(i, "Nilai i")
  #print(j, "Nilai j")
  for j in range(count,i,-1): #proses perulangan yang akan terjadi dari nilai count sampai i dengan langkah mundur
    if data[i:j] in kamus: #proses seleksi yang akan menjalankan program dalam if jika data dengan indeks[i:j] berada dalam kamus
      getIndex = kamus.index(data[i:j]) #proses inisialisasi getIndex diisikan dengan kamus pada index dalam data[i:j]
      #print(getIndex)
      if j==count: #penyeleksian jika j == count maka akan menjalankan program dalam if
        kamus.append(None) #memasukkan nilai None ke kamus
        hasilencode.append((getIndex+1,None)) #memasukkan hasilencode dengan data getIndex dijumlahkan 1, dan diisi juga dengan None

      else : #penyeleksian jika j != 0 maka akan menjalankan program dalam else
        kamus.append(data[i:j]+data[j]) #memasukkan nilai data[i:j] dengan data[j] kedalam kamus
        #print(data[i:j]+ " ini I dan J" + " " + str(i))
        #print(data[j] + " ini J aja" + " " + str(j))
        hasilencode.append((getIndex+1,data[j])) #memasukkan hasilencode dengan data getIndex dijumlahkan 1, dan diisi juga dengan data[j]

        #i=i+2
      j=j+1#memasukkan nilai j dengan j ditambah 1
      break #perintah khusus yang dipakai untuk memaksa sebuah perulangan berhenti sebelum waktunya

    else: #penyeleksian jika data dengan indeks[i:j] tidak berada dalam kamus maka akan menjalankan program dibawah ini
      if len(data[i:j])==1: #penyeleksian kondisi jika panjang data[i:j] sama dengan 1 maka akan menjalankan program dibawahnya
        kamus.append(data[i:j]) #memasukkan nilai data[i:j] ke dalam kamus
        hasilencode.append((0,data[i:j])) #memasukkan nilai data[i:j] dengan angka 0 di awal untuk data hasilencode

        break #perintah khusus yang dipakai untuk memaksa sebuah perulangan berhenti sebelum waktunya

      else: #penyeleksian kondisi jika panjang data[i:j] tidak sama dengan 1 maka akan menjalankan program dibawahnya

        continue #perintah khusu yang dipakai untuk melewati 1 proses perulangan saja
print('Dictionary:' +str(kamus)) #mencetak Dictionary dan dengan nilai kamus yang telah dikonversi menjadi string
print('\nData Stream: '+str(data)) #mencetak Data Stream dan dengan nilai data yang telah dikonversi menjadi string dengan baris baru
print('\nHasil Encode :'+str(hasilencode)) #mencetak Hasil Encode dan dengan nilai hasilencode yang telah dikonversi menjadi string dengan baris baru
```



> Analisa dari program di atas:

Dari program diatas dapat diketahui bahwa terdapat dua buah kondisi dimana jika data sudah ada dalam kamus atau belum, dalam kondisi awal pastinya data tidak akan memiliki data dalam kamus sehingga akan memasuki bagian else dimana data tidak ada dalam kamus sehingga akan menambahkan data baru ke dalam kamus dengan cara memanfaatkan append dimana fungsi ini digunakan untuk memasukkan data yang berupa list. Kemudian karena belum ada dalam kamus sehingga dalam hasil encode untuk nilai index akan diisikan dengan nilai 0, kemudian untuk hurufnya adalah data yang sudah baru dimasukkan ke dalam kamus yaitu B, lalu untuk data selanjutnya yaitu A maka karena A juga sama belum ada dalam kamus maka akan dimasukkan ke dalam kondisi program else, lalu melakukan proses yang sama seperti data B, dengan hasil encodenya adalah 0 dikarenakan tidak ada keterangan index yang didapatkan dalam ka
mus, lalu setelah koma akan diisikan dengan data huruf yang baru saja dimasukkan ke kamus yaitu A. Kemudian untuk data selanjutnya yaitu B karena B sendiri sudah ada datanya dalam kamus maka akan masuk ke dalam kondisi if pertama kemudian akan ada proses untuk pengambilan index ke berapa kamus yang datanya adalah B, kemudian setelah didapatkan maka akan dimasukkan ke dalam variavel getIndex, lalu dalam kode if ini terdapat if kembali namun untuk kondisi akhir saja yaitu dimana jika j sudah sama dengan count yaitu jumlah data, karena data B ini masih belum akhir dari data maka akan masuk kondisi else. Dimana akan ada proses pemasukkan data baru ke kamus yaitu data dengan index ke i yaitu index kedua yaitu B dan data dengan index ke j yaitu index ketiga dengan A(perlu diingat bahwa index akan dimulai dari 0). Kemudian setelah dimasukkan data baru ke kamus maka akan ada proses pemasukkan nilai ke hasil encode yaitu nilai getIndex ditambahkan 1 karena nilai index berawal dari 0 sedangkan index yang kita mau berawal dari 1, sehingga perlu ada proses penambahan 1. Lalu ditambahkan data dengan index ke j yaitu data B. Semua proses ini akan terus berulang sampai nantinya i sudah melewati banyaknya data.
