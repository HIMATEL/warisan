### Pendahuluan

Dalam usaha untuk mengirimkan data dari satu tempat ke tempat lain, seringkali terjadi ketidakseimbangan antara jumlah data yang akan dikirim dengan kapasitas jaringan yang tersedia. Akibatnya, kualitas layanan dalam sistem pengiriman data menjadi kurang memadai. Salah satu cara untuk mengatasi situasi ini adalah dengan melakukan kompresi data sebelum pengiriman, sehingga beban pada jaringan dapat berkurang. Teknik kompresi data dapat diterapkan jika terdapat pola berulang dalam data, di mana pola-pola ini dapat diwakili dengan singkat pada awalnya dan kemunculan selanjutnya hanya merujuk pada pola awal tersebut. Terdapat berbagai teknik kompresi data yang sudah dikembangkan, dan pada kesempatan ini akan diterapkan metode kompresi data yang dikenal sebagai Run Length Encoding (RLE). RLE merupakan teknik kompresi data yang mengompresikan pola yang muncul secara berurutan. Contoh penggunaan RLE adalah dengan mengganti setiap rangkaian angka 0 dengan 2 karakter, yang menunjukkan bahwa kompresi sedang dilakukan, diikuti dengan jumlah angka 0 dalam rangkaian tersebut. Semakin banyak angka 0 yang berurutan, maka ukuran file hasil kompresi akan semakin efisien.

Berdasarkan pada program RLE Encoder sebelumnya, algoritma program RLE Decoder seharusnya tidak jauh berbeda dari program sebelumnya. Di program Encoder mahasiswa sudah dapat membuat program untuk mencari suatu simbol yang akan diencode (atau dalam kasus sebelumnya mencari angka 0), kemudian saat simbol tersebut akan dicek ke simbol berikutnya apakah masih berulang atau tidak. Jika ditemukan simbol berulang maka simbol tersebut akan dihitung, sebaliknya jika tidak ada perulangan maka program hanya mencetak ulang simbol dari data stream.

Pada bagian decoder ini, mahasiswa diminta untuk bisa membuat program yang dapat mencari simbol 0. Kemudian setelah simbol 0 ditemukan, program harus bisa melihat angka didepan simbol 0 tersebut. Angka tersebut merupakan jumlah perulangan dari angka 0 yang akan didecode. Contoh jika didapatkan;

**datastream** = 1 2 3 0 5 8 9

maka setelah ditemukan angka 0, output harus bisa mencetak ulang angka 0 sebanyak angka didepannya yaitu 5. Sehingga data output yang dihasilkan menjadi;

**output** = 1 2 3 0 0 0 0 0 8 9

1. Buatlah program untuk melakukan decoding RLE dengan mencari angka 0 lalu mengulang akan 0 tersebut sebanyak jumlah angka yang ada setelahnya.



```python
c = 0 #berfungsi sebagai urutan karakter pada variabel datastream yang dimulai dari 0
count = 0 #berfungsi sebagai hasil hitung untuk karakter 0 pada variabel datastream
jml = 0 #berfungsi sebagai tempat menyimpan jumlah 0

datastream = '451325345077623145'#berfungsi sebagai inisialisasi variabel datastream

for x in range(len(datastream)): #berfungsi sebagai pengulangan/loop berdasarkan ‘x’ yang berada di rentang jumlah karakter yang ada pada datastream
    if c != len(datastream) : #befungsi sebagai ketika c tidak berada dalam jumlah karakter variabel datastream
        if datastream[c] !='0':#berfungsi sebagai intrupsi percabangan untuk menentukan keadaan jika dalam keadaan data pada posisi bernilai tidak sama dengan 0
            if datastream[c-1] == '0': #'0’ berfungsi sebagai intrupsi percabangan untuk menentukan keadaan jika dalam keadaan data c -1 pada posisi bernilai 0
                jml = int (datastream[c]) #berfungsi sebagai deklarasi isi variabel jml terdapat datastream[c] dengan tipe data integer
                for y in range (jml):#berfungsi sebagai pengulangan/loop berdasarkan ‘y’ yang berada di rentang dari variabel jml
                    print('0', end='') #berfungsi sebagai untuk mencetak hasil dari perintah for y in range (jml):
                c=c+1 #proses increment untuk variable c, yang berfungsi untuk berpindah ke urutan karakter selanjutnya pada datastream, saat proses identifikasi karakter
        print(datastream[c],end='') #berfungsi sebagai untuk mencetak hasil dari perintah if datastream[c] != '0':
    c=c+1 #proses increment untuk variable c, yang berfungsi untuk berpindah ke urutan karakter selanjutnya pada data datastream, saat proses identifikasi karakter
```

2. Buatlah program untuk melakukan decoding RLE dengan mencari angka apapun yang
mengulang. Cetak angka tersebut sebanyak perulangan yang diminta (jumlah perulangan pada program ini ditandai dengan tanda minus).



```python
c = 0 #berfungsi sebagai urutan karakter pada variabel datastream yang dimulai dari 0
count = 0 #berfungsi sebagai hasil hitung untuk karakter 0 pada variabel datastream
jml = 0 #berfungsi sebagai tempat menyimpan jumlah integer
d = '-' #berfungsi sebagai inisialisasi tanda minus

datastream = '65-2454-454542-30-312346'#berfungsi sebagai inisialisasi variabel datastream

for x in range(len(datastream)): #berfungsi sebagai pengulangan/loop berdasarkan ‘x’ yang berada di rentang jumlah karakter yang ada pada datastream
    if c != len(datastream) : #befungsi sebagai ketika c tidak berada dalam jumlah karakter variabel datastream
        if datastream[c] ==d:#berfungsi sebagai intrupsi percabangan untuk menentukan keadaan jika dalam keadaan data pada posisi bernilai sama dengan tanda minus
            jml = int (datastream[c+1]) # sebagai deklarasi isi variabel jml terdapat datastream[c] dengan tipe data integer
            for y in range (jml-1):#berfungsi sebagai pengulangan/loop berdasarkan ‘y’ yang berada di rentang dari variabel jml
                print(datastream[c-1],end='') #berfungsi sebagai untuk mencetak hasil dari perintah for y in range(jml-1):
            c=c+2 #proses increment untuk variable c, yang berfungsi untuk berpindah ke urutan karakter selanjutnya pada datastream, saat proses identifikasi karakter
        if datastream[c] !=d:# berfungsi sebagai intrupsi percabangan untuk menentukan keadaan jika dalam keadaan data pada posisi bernilai tidak sama dengan tanda minus
            print(datastream[c],end='') #berfungsi sebagai untuk mencetak hasil dari perintah if datastream[c] !=d:
        c=c+1 #proses increment untuk variable c, yang berfungsi untuk berpindah ke urutan karakter selanjutnya pada datastream, saat proses identifikasi karakter
```
