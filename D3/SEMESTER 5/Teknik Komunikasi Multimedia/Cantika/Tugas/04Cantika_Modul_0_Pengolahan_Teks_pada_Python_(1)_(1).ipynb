```python
string = 'Politeknik Negeri Bandung'
variabel = [None]*30
for i in range(len(string)):
  variabel[i]=string[i]
print(variabel[0])
print(variabel[1])
print(variabel[2])
print(variabel[3])
```

a. Program ini digunakan untuk menyimpan setiap huruf dari kalimat 'Politeknik Negeri Bandung' ke dalam sebuah daftar (list), lalu menampilkan beberapa huruf pertama dari daftar tersebut.

b. variabel = [None]*30
Membuat daftar (list) bernama variabel yang berisi 30 elemen, semuanya diisi dengan None (nilai kosong). Daftar ini dibuat cukup besar untuk menampung semua karakter dalam kalimat . Tempat ini akan kita isi dengan huruf-huruf dari string. **for i in range(len(string)):**
Baris ini memulai sebuah pengulangan (loop) yang akan berjalan sebanyak jumlah huruf dalam string. Setiap kali loop ini berjalan, i adalah angka yang mewakili posisi dari setiap huruf. **variabel[i] = string[i]**
Di sini, kita memasukkan setiap huruf dari string ke dalam daftar variabel sesuai dengan urutannya. Misalnya, huruf pertama 'P' akan masuk ke posisi pertama dalam variabel. **print(variabel[0])**
Baris ini mencetak huruf pertama dalam daftar variabel, yaitu 'P'.
**print(variabel[1])**
Baris ini mencetak huruf kedua dalam daftar variabel, yaitu 'o'.
**print(variabel[2])**
Baris ini mencetak huruf ketiga dalam daftar variabel, yaitu 'l'.
**print(variabel[3])**
Baris ini mencetak huruf keempat dalam daftar variabel, yaitu 'i'.
