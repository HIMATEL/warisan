### Pendahuluan

**Elias Gamma Coding**

Metode ini dikembangkan oleh Peter Elias. Dengan teknik pengkodean:
1. Tulis dalam biner
2. Jumlah bit pada proses no.1 dikurang satu untuk menentukan jumlah nol (zeros) yang
harus ditambahkan di muka angka tersebut.
Contoh ada sebuat bilangan 11:
1. Langkah pertama adalah konversi 11 menjadi biner, yakni 1011.
2. Cari jumlah bit/pangkat tertinggi pada proses pertama, lalu kurangi 1.

Gunakan jumlah tersebut untuk menambahkan angka 0 didepan biner yang terdapat pada proses pertama.
Hasilnya adalah 0001011.

Sehingga jika didapatkan deretan angka 11, 15, dan 27 maka dengan elias gamma code deretan angka tersebut dapat diubah menjadi 00010110001111000011011.

Fungsi dari elias gamma code ini adalah pengguna dapat mengirim banyak angka berjajar sekaligus, sebab jika kita hanya mengubah bilangan tersebut menjadi biner maka akan sulit melakukan decode.

Sebagai contoh tiga bilangan yang terdapat pada contoh di atas. Jika data
tersebut dikirim 1011111111011 saja, maka akan ada banyak variasi kombinasi biner untuk mendapatkan nilai desimalnya.

**Elias Delta Coding**

Seperti Elias Gamma, kode ini ditemukan oleh Peter Elias. Kode ini menerapkan juga metode
pada gamma coding, terutama di bagian kepala. Tekniknya adalah sebagai berikut:
1. Cari pangkat tertinggi binernya, misalnya desimal 11 jika dibinerkan menjadi 1011 dimana
pangkat tertingginya adalah 3. Jadi N’=3.
2. Gunakan Gamma Coding untuk mengkodekan bilangan N dimana N=N’+1. Jadi untuk
kasus desimal 11 maka kita harus membuat Gamma Coding dari 4 yaitu 00100.
3. Tambahkan sisa N’ binary pada hasil no.2. Jadi diperoleh jawaban 00100011.
Jika diberikan deretan angka 35, 101, dan 112, maka akan didapatkan output dengan metode Elias Delta Code sebagai berikut:

00110000110011110010100111110000

```python
#tuliskan fungsi program pada tiap barisnya

def elliasGamma(input): #untuk apa perintah def? Untuk membuat sebuah function, sehingga akan menghemat kode disaat akan digunakan berulang ulang
 gammaResult = "" #apa fungsi variabel gammaResult? Untuk menyimpan sebuah nilai akhir dari encode eliassgamma
 biner = format(input,'b') #untuk apa perintah format? untuk merubah bilangan desimal menjadi biner

 for n in range(len(biner)-1): #apa tujuan dari for ini? untuk mengulangi proses penambahan nol dengan banyaknya bit dari data dikurang 1
  gammaResult = '0' + gammaResult

 gammaResult = gammaResult + biner

 return gammaResult #untuk apa perintah return? untuk mengembalikan nilai, dalam program ini maka gammaResult nantinya akan dikembalikkan nilainya dan masuk ke variabel yang memanggil function

#----------------------------------------------

def elliasDelta(input): #untuk apa perintah def? Untuk membuat sebuah function, sehingga akan menghemat kode disaat akan digunakan berulang ulang
 biner = format(input,'b') #untuk apa perintah format? Untuk menyimpan sebuah nilai akhir dari encode eliassgamma
 n = len(biner) #apa fungsi variabel n, untuk menampung nilai dari panjangnya bit biner
 #print(n)
 gamma = elliasGamma(n) #apa fungsi variabel gamma, untuk menampung nilai gammaResult yang akan dikembalikan
 deltaResult = gamma + biner[1:] #apa fungsi variabel deltaResult? untuk apa biner[1:], dataResult digunakan untuk menampung nilai hasil akhir eliasdelta encode, lalu biner[1:] untuk hanya mengambil nilai diatas indeks 1 saja

 return deltaResult #untuk apa perintah return? untuk mengembalikan nilai, dalam program ini maka deltaResult nantinya akan dikembalikkan nilainya dan masuk ke variabel yang memanggil function

#----------------------------------------------

data = [23, 35, 36, 44, 45] #untuk apa variabel data? untuk menampung nilai yang akan di encode
out_gamma = ''
out_delta = ''

for i in data:
 out_gamma = out_gamma + ' ' + elliasGamma(i) #untuk apa variabel out_gamma? untuk menampung nilai akhir dari keseluruhan data yang sudah diencode dengan metode elliasgamma
 #print(i)
 out_delta = out_delta + ' ' + elliasDelta(i) #untuk apa variabel out_delta? untuk menampung nilai akhir dari keseluruhan data yang sudah diencode dengan metode elliasdelta

print('input gamma: \n\t',data)
print('output gamma: \n\t',out_gamma)
print('input delta: \n\t',data)
print('output delta: \n\t',out_delta)
```
